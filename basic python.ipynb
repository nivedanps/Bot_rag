```python
from os.path import exists
from typing_extensions import Required
''''' an auto encoder is a type of neural network architecture that is used in deep learning for unsupervised learning and dimensionality reduction task .It is
primararily designed to learn efficient representation of data by encoding it into a latent space and the encoding it back to the original data space auto encoders
are widely used for various application such as data compression,denoising,feature learning and anamoley detection'''


Architecture of auto encoder two main parts
encoder and decoder


'''architecture of auto encoder
Encoder-encoder networks takes the input data in maps it to lower dimensional latent space which
 is often called as encoding or latent space .this encoding is typically compressed representationof the input data .
 the encoder can consists of 1 or more hidden layers and uses activation functions


Decoder-  decoder network takes the encoded represetation and attempt to reconstruct the orginal input
data from it .Like the encoder the decoder can also have one or more hidden layer and activation layer .


 The main objective of the auto encoder is to minize the reconstruction which measures how well the decoder can reconstruct
the input data from the encoded representation. common less function used for this purpose includes MSE(mean Squared error )
or binary cross entropy depending on the type of the data (continous or binary and the specific tasks



Training -auto encoder are trained in an unsupervised manner which means they donot require labbeled data. the training
process involves feeding input data into the encoder and encoding  it  to a lower dimensional representation and then decoded
it the original data the differnece the input and the reconstructed the output is to compute the loss
which is minized during the traning using the optimization technique  like gradiant decent)'''


Varitions in aut encoder
There are several variations of autoencoder each designed for specific task for data types.
some common types includes
1-denoising auto encoder-Trained to remove noise data by corrupting the input and reconstructing the clean data.
2-sparse Auto encoder- encourages the network to learn sparse representation whcih can be useful for feautre learning and dimensionality reduction.
3-variational auto encoder(VAE)-introduces proboabilistic elmensts in the latent space allowing for the  generation of new data samples and improve data compression.
4-convolutional Auto Encoder-specifically designed for image data using convolutional layer in the encoder and decoder.
5-recurrent Auto encoder-suitable for sequential data such as the time series by using recurrent layer in the architecture.


Now if we consider the numerical aspect of an autoencoder here is how it works
1-input data-Autoencoder takes numerical data as input this data could be anything from image(pixel values,audio) amplitude values
 over time text(both tokenization word embedding or any other form of numerical representation)
2-encoder-the encoder network recieves this numerical data and maps it to an lower dimensional numerical representation
for example in the case of image data the encoder might transform pixel values into a reduced set of numerical features
 that captures essential information.
3-latent space-compressed lower-dimensional and representation of high -dimensional input data.
4-decoder-the decoder takes the reduced numerical representation and attempts to reconstruct the original input data it does so by mapping the
reduced representation back to the original numerical format.
5-loss function-During traning the differnece between the input numerical data and the reconstructed data is measured using numerical loss
function the model adjusts its weights and biases from numerical optimization techniqueto minimize this loss


In Summary auto encoders are versatile tool in deep learning for learning compact representations of data and have numerious
representation in various domain working with both numerical and non numerical data


1-neural network-it generates same output
2-auto encoder-it generates different output based on the input,
3-variatinal auto encoder-it is same as AE it just provides better image.
4-Generative Model-it randomnly generate Images based on user Requirment


normal auto encoder                                                  |                 Variational autoencoder
why does this exist?                                                              why does this exists
what does this aptimize                                                           learns to generate new data
minimize Reconstruction loss                                                      what does this optimize
cannot generate new iamge                                                         Minimize reconsturction loss and latent loos


Generative model                                                      |               Variational autoencoder
how does this learns to genarative new data                                       consist of an encoder and decoder
consists of a generator and a discriminator networks
'''
'''




```
